````python
# Rainfall data (in mm) and runoff coefficient for Karun river basin
rainfall_karun = [200, 250, 300, 150, 400]  # Monthly rainfall data
runoff_coefficient = 0.6  # 60% of rainfall becomes runoff
# Calculate monthly runoff
runoff_karun = [rain * runoff_coefficient for rain in rainfall_karun]
# Print monthly runoff data
print("Monthly runoff in Karun river basin (in mm):")
for i, runoff in enumerate(runoff_karun, start=1):
    print(f"Month {i}: {runoff} mm")
# Calculate total annual runoff
total_runoff = sum(runoff_karun)
print(f"Total annual runoff: {total_runoff} m""salaries (2).csv")
````

````python
# Rainfall data (in mm) and runoff coefficient for Karun river basin
rainfall_karun = [200, 250, 300, 150, 400]  # Monthly rainfall data
runoff_coefficient = 0.6  # 60% of rainfall becomes runoff
# Calculate monthly runoff
runoff_karun = [rain * runoff_coefficient for rain in rainfall_karun]
# Print monthly runoff data
print("Monthly runoff in Karun river basin (in mm):")
for i, runoff in enumerate(runoff_karun, start=1):
    print(f"Month {i}: {runoff} mm")
# Calculate total annual runoff
total_runoff = sum(runoff_karun)
print(f"Total annual runoff: {total_runoff} mm")
````

# While

````python
i = 0
while i < 5:
    print("Iteration:", i)
    i += 1
````

````python
fruits = ["apple", "banana", "cherry"]
i = 0
while i < len(fruits):
    print(fruits[i], end=" / ")
    i += 1
````

````python
thistuple = ("apple", "banana", "cherry")
i = 0
while i < len(thistuple):
    print(thistuple[i])
    i += 1
````

````python
fruits = ["apple", "banana", "cherry"]
i = 0
while i < len(fruits):
    print(fruits[i])
    if fruits[i] == "banana":
        break
    i += 1
````

````python
thislist = ["apple", "banana", "cherry"]
i = 0
while i < 3:
    print(thislist[i])
    i += 1
````

````python
thistuple = ("apple", "banana", "cherry")
i = 0
while i < len(thistuple):
    print(thistuple[i])
    i += 1
````

````python
list1 = ["a", "b", "c"]
list2 = [1, 2, 3]
i = 0
while i < len(list2):
    list1.append(list2[i])
    i += 1

print(list1)
````
