```python
import matplotlib.pyplot as plt

# Dummy data
x = [1, 2, 3, 4, 5]
y1 = [10, 20, 25, 30, 40]
y2 = [3, 15, 28, 32, 43]
y3 = [6, 35, 40, 25, 30]

# Draw vertical and horizontal dashed lines from the point to the axes
x_val, y_val1, y_val2, y_val3 = 2, 20, 15, 35
plt.axvline(x=x_val, color='black', linestyle='dotted', linewidth=2)
plt.axhline(y=y_val1, color='black', linestyle='dotted', linewidth=2)
plt.axhline(y=y_val2, color='black', linestyle='dotted', linewidth=2)
plt.axhline(y=y_val3, color='black', linestyle='dotted', linewidth=2)

# Annotate the point with its coordinates
plt.text(x_val + 0.1, y_val1 - 1.7, f'Coordinates: ({x_val:.2f}, {y_val1:.2f})', fontsize=10, color='orange')
plt.text(x_val + 0.1, y_val2 - 1.7, f'Coordinates: ({x_val:.2f}, {y_val2:.2f})', fontsize=10, color='purple')
plt.text(x_val + 0.1, y_val3 - 1.7, f'Coordinates: ({x_val:.2f}, {y_val2:.2f})', fontsize=10, color='gray')

# Plotting
plt.plot(x, y1, color='green', linestyle='solid', linewidth=2, marker='o', markersize=6, label='Line 1')
plt.plot(x, y2, color='red', linestyle='dashed', linewidth=2, marker='s', markersize=6, label='Line 2')
plt.plot(x, y3, color='blue', linestyle='dashdot', linewidth=2, marker='D', markersize=6, label='Line 3')

plt.title("Multiple Lines Plot", fontname='Times New Roman', fontsize=16)
plt.xlabel("X-axis", fontname='Times New Roman', fontsize=14)
plt.ylabel("Y-axis", fontname='Times New Roman', fontsize=14)

plt.xticks(x, fontname='Times New Roman', fontsize=12)
plt.yticks(fontname='Times New Roman', fontsize=12)

plt.xlim(0, 6)
plt.ylim(0, 50)

plt.legend(loc='best')
# 'best', 'left', 'right', 'center', 'lower left', 'lower right', 'upper left', 'upper right'
plt.savefig('plot.png', dpi=300)

plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt

# Dummy data
categories = ['A', 'B', 'C', 'D']
values = [10, 20, 15, 25]

# Plotting
plt.bar(categories, values, color=['orange', "red", "blue", "black"])

plt.title("Basic Bar Plot", fontname='Times New Roman', fontsize=16)
plt.xlabel("Categories", fontname='Times New Roman', fontsize=14)
plt.ylabel("Values", fontname='Times New Roman', fontsize=14)

plt.xticks(categories, fontname='Times New Roman', fontsize=12)
plt.yticks(values, fontname='Times New Roman', fontsize=12)

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Dummy data
categories = ['A', 'B', 'C', 'D']
values1 = [10, 20, 15, 25]
values2 = [15, 25, 20, 30]

# Plotting
x = np.arange(len(categories))
width = 0.35
# Note: If we remove x - width/2 and x + width/2, the bars will overlap!
plt.bar(x - width/2, values1, width, color='blue', label='Group 1')
plt.bar(x + width/2, values2, width, color='green', label='Group 2')

plt.title("Grouped Bar Plot", fontname='Times New Roman', fontsize=16)
plt.xlabel("Categories", fontname='Times New Roman', fontsize=14)
plt.ylabel("Values", fontname='Times New Roman', fontsize=14)

plt.xticks(ticks=x, labels= categories, fontname='Times New Roman', fontsize=12)
plt.yticks(values, fontname='Times New Roman', fontsize=12)

plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt

# Data for the pie chart
labels = ['Category A', 'Category B', 'Category C', 'Category D']
sizes = [25, 35, 20, 20]
explode = (0, 0.1, 0, 0)  # Explode the slice for Category B

# Data for the exploded slice composition (Category B)
sub_labels = ['Component 1', 'Component 2', 'Component 3']
sub_sizes = [40, 35, 25]  # Percentages sum to 100
plt.figure(figsize=(12, 6))

# Pie chart with the exploded slice
plt.subplot(1, 2, 1)
plt.pie(sizes, labels=labels, explode=explode, autopct='%1.1f%%', shadow=True,textprops={"fontname":"Times New Roman", "fontsize":14}, startangle=90)
plt.title('Pie Chart with Exploded Slice', fontname='Times New Roman', fontsize=16)

# Separate bar chart showing the composition of the exploded slice
plt.subplot(1, 2, 2)
bars = plt.bar(sub_labels, sub_sizes, color=['blue', 'green', 'red'])
plt.title('Composition of Exploded Slice', fontname='Times New Roman', fontsize=16)
plt.xlabel('Components', fontname='Times New Roman', fontsize=14)
plt.ylabel('Percentage', fontname='Times New Roman', fontsize=14)

plt.xticks(x, fontname='Times New Roman', fontsize=12)
plt.yticks(fontname='Times New Roman', fontsize=12)

plt.tight_layout()
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Dummy data
data = [12, 15, 13, 14, 15, 8, 16, 14, 18, 14, 13, 12, 23, 15, 16, 17]

flierprops = dict(marker='D', markerfacecolor='red', markersize=8)

# Plotting
sns.boxplot(data, color='orange', flierprops=flierprops)
plt.title("Basic Box Plot", fontname='Times New Roman', fontsize=16)
plt.yticks(range(5, 25, 2), fontname='Times New Roman', fontsize=12)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D 

# Generate random data
np.random.seed(42)  # for reproducibility

n = 100  # number of points
x = np.random.rand(n)
y = np.random.rand(n)
z = np.random.rand(n)

# Create a figure and a 3D axis
fig = plt.figure(figsize=(10,6))
# ax = fig.add_subplot(111, projection='3d')
ax = plt.axes(projection='3d')

# Plot the 3D scatter plot
ax.scatter(x, y, z, c='r', marker='o')

# ax.view_init(elev=20, azim=135)
ax.view_init(elev=0, azim=0)

# Set title and axis labels
ax.set_title('3D Scatter', fontname='Times New Roman', fontsize=16)
ax.set_xlabel('X Axis', fontname='Times New Roman', fontsize=14)
ax.set_ylabel('Y Axis', fontname='Times New Roman', fontsize=14)
ax.set_zlabel('Z Axis', fontname='Times New Roman', fontsize=14)

# Display the plot
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Create data for a 3D surface
x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)
x, y = np.meshgrid(x, y)
r = np.sqrt(x**2 + y**2)  
z = np.where(r == 0, 1.0, np.sin(r)/r)

# Create a figure and a 3D axis
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

# Plot the surface
# surf = ax.plot_surface(x, y, z, cmap='viridis', edgecolor='none')
# surf = ax.plot_surface(x, y, z, cmap='inferno', edgecolor='none')
surf = ax.plot_surface(x, y, z, cmap='plasma', edgecolor='none')
# surf = ax.plot_surface(x, y, z, cmap='magma', edgecolor='none')
# surf = ax.plot_surface(x, y, z, cmap='cividis', edgecolor='none')

# Add a color bar to map values to colors
fig.colorbar(surf, shrink=0.7, aspect=8)

# Set title and axis labels
ax.set_title('3D Surface Plot', fontname='Times New Roman', fontsize=16)

ax.set_xlabel('X Axis', fontname='Times New Roman', fontsize=10)
ax.set_ylabel('Y Axis', fontname='Times New Roman', fontsize=10)
ax.set_zlabel('Z Axis', fontname='Times New Roman', fontsize=10)

# Display the plot
plt.show()
```
