## if & else

```python
a = 33
b = 200
if b > a:
  print("b is greater than a")
```

```python
a = 33
b = 33
if b > a:
  print("b is greater than a")
elif a == b:
  print("a and b are equal")
```

```python
a = 200
b = 33
if b > a:
  print("b is greater than a")
elif a == b:
  print("a and b are equal")
else:
  print("a is greater than b")
```

```python
a = 200
b = 33
c = 500
if a > b and c > a:
  print("Both conditions are True")
```

```python
a = 200
b = 33
c = 500
if a > b or a > c:
  print("At least one of the conditions is True")
```

```python
x = 41
if x > 10:
  print("Above ten,")
  if x > 20:
    print("and also above 20!")
  else:
    print("but not above 20.")
```

```python
thistuple = ("apple", "banana", "cherry")
if "apple" in thistuple:
  print("Yes, 'apple' is in the fruits tuple")
```

```python
# If-elif example
temperature = 30
if temperature > 40:
    print("It's very hot!")
elif temperature > 30:
    print("It's hot.")
else:
    print("It's pleasant.")
```

```python
# Get rainfall input from the user
rain = float(input("Enter rainfall amount (in mm): "))
# Check rainfall status
if rain > 50:
    print("Heavy rainfall! Be careful.")
else:
    print("Light rainfall.")
```

```python
# Get temperature input from the user
temp = float(input("Enter temperature (in °C): "))
# Check temperature status
if temp > 40:
    print("It's very hot! Stay indoors.")
elif temp > 30:
    print("It's hot. Drink plenty of water.")
elif temp > 20:
    print("The weather is pleasant. Enjoy!")
else:
    print("It's cold. Wear warm clothes.")
```

```python
# Get two numbers from the user
num1 = float(input("Enter the first number: "))
num2 = float(input("Enter the second number: "))
# Compare the two numbers
if num1 > num2:
    print("The first number is greater.")
elif num1 < num2:
    print("The second number is greater.")
else:
    print("Both numbers are equal.")
```

```python
# Get user inputs
rainfall_karun = float(input("Enter rainfall in Karun basin (in mm): "))
runoff_coefficient = float(input("Enter runoff coefficient (between 0 and 1): "))
# Calculate runoff
runoff_karun = rainfall_karun * runoff_coefficient
# Check the runoff condition
if runoff_karun > 300:
    print("High runoff in Karun river! Risk of flooding.")
elif runoff_karun > 200:
    print("Moderate runoff in Karun river. Be alert.")
elif runoff_karun > 100:
    print("Low runoff in Karun river. Normal conditions.")
else:
    print("Very low runoff in Karun river. Drought conditions.")
```
