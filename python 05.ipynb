```python
def calculate_circle(x):
    w = 3.14159 * x ** 2
    z = 2 * 3.14159 * x
    return w, z 

w, z = calculate_circle(5)

print(f"Area: {w: 0.2f}, Circumference: {z: 0.2f}")
```

```python
def celsius_to_fahrenheit(celsius):
    fahrenheit = (celsius * 9/5) + 32
    return fahrenheit

temp_c = 25
temp_f = celsius_to_fahrenheit(temp_c)
print(f"{temp_c}°C = {temp_f:.1f}°F") 
```

```python
def meter_converter(meters):
    feet = meters * 3.28084
    inches = meters * 39.3701
    return feet, inches

ft, inch = meter_converter(1.8)
print(f"Height: {ft:.2f} ft / {inch:.2f} in") 
```

```python
def generate_username(first, last, birth_year):
    return f"{first[:3]}{last[:2]}{birth_year % 100}"

username = generate_username("Aliakbar", "Noroztavvali", 1999)
print(f"Your username: {username}") 


def calculate_tip(bill_amount, tip_percent=10):
    tip = bill_amount * (tip_percent / 100)
    total = bill_amount + tip
    return tip, total
tip, total = calculate_tip(280000, 15)
print(f"Tip: {tip:.0f} Toman | Total: {total:.0f} Toman")
```

```python
def is_strong_password(password):
    
    has_upper = any(c.isupper() for c in password)
    has_lower = any(c.islower() for c in password)
    has_digit = any(c.isdigit() for c in password)
    length_ok = len(password) >= 8
        # Show password weakness reasons
    if not all([length_ok, has_upper, has_lower, has_digit]):
        reasons = []
        if not length_ok:
            reasons.append("Password must be at least 8 characters long")
        if not has_upper:
            reasons.append("Missing uppercase letter")
        if not has_lower:
            reasons.append("Missing lowercase letter")
        if not has_digit:
            reasons.append("Missing digit")
        return False, reasons
    return True, "Password is strong"
# Testing
result, msg = is_strong_password("python32")
print(f"Result: {result} → {msg}")
```
